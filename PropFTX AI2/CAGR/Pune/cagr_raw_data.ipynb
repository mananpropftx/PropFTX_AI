```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

data = {
    "location": ["Airport Road"] * 5,
    "year": [2024, 2024, 2024, 2024, 2025],
    "quarter": ["Jan-Mar", "Apr-Jun", "Jul-Sep", "Oct-Dec", "Jan-Mar"],
    "avg_price": [6858.92, 6494.38, 6384.00, 6532.07, 6738.23],
    "max_price": [8407.30, 7960.46, 7825.17, 8006.66, 8259.36],
    "min_price": [5310.54, 5028.30, 4942.84, 5057.48, 5217.10]
}
df = pd.DataFrame(data)

df_airport = df[df['location'] == 'Airport Road']

df_airport['time'] = df_airport['year'].astype(str) + " " + df_airport['quarter']

x = np.arange(len(df_airport))

min_prices = df_airport['min_price']
avg_prices = df_airport['avg_price'] - min_prices
max_prices = df_airport['max_price'] - df_airport['avg_price']

plt.figure(figsize=(12, 6))

plt.bar(x, min_prices, label='Min Price', color='skyblue')
plt.bar(x, avg_prices, bottom=min_prices, label='Avg Price', color='lightgreen')
plt.bar(x, max_prices, bottom=df_airport['avg_price'], label='Max Price', color='salmon')

plt.plot(x, df_airport['avg_price'], color='black', marker='o', linewidth=2, label='Avg Price Trend')

plt.xticks(x, df_airport['time'], rotation=45)
plt.ylabel("Price")
plt.title("Price Trends for Airport Road")
plt.legend()
plt.tight_layout()
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```
