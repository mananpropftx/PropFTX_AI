```python
import pandas as pd
import numpy as np

df = pd.read_csv('C:\\Users\\Administrator\\Downloads\\PropFTX AI2\\PropFTX AI2\\Prediction\\Mumbai\\Mumbai Property Data.csv')

df['Locality'] = df['Locality'].str.strip().str.lower()

df['Average Price'] = df['Average Price'].str.replace(',', '', regex=False)  
df['Average Price'] = df['Average Price'].replace('-', np.nan)             
df['Average Price'] = pd.to_numeric(df['Average Price'], errors='coerce')   

df = df.dropna(subset=['Average Price'])

start_df = df[df['Quarter'] == 'Jul-Sep 2018'][['Locality', 'Average Price']].copy()
end_df = df[df['Quarter'] == 'Jul-Sep 2024'][['Locality', 'Average Price']].copy()

start_df = start_df.rename(columns={'Average Price': 'Start_Price'})
end_df = end_df.rename(columns={'Average Price': 'End_Price'})

merged = pd.merge(start_df, end_df, on='Locality', how='inner')
merged['CAGR'] = ((merged['End_Price'] / merged['Start_Price']) ** (1/6)) - 1
merged['CAGR (%)'] = (merged['CAGR'] * 100).round(2)

print(merged[['Locality', 'Start_Price', 'End_Price', 'CAGR (%)']])
merged.to_csv('Mumbai_cagr_values.csv')
```

```python
negative_cagr_df = merged[merged['CAGR'] < 0]
negative_cagr_df = negative_cagr_df[['Locality', 'Start_Price', 'End_Price', 'CAGR (%)']]
negative_cagr_df.to_csv('Mumbai_negative_cagr_localities.csv', index=False)
print("Saved localities with negative CAGR to 'Mumbai_negative_cagr_localities.csv'")
```
